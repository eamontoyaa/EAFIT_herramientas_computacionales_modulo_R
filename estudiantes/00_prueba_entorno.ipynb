```r
# Prueba espacial opcional (puede instalar paquetes faltantes)
spatial_pkgs <- c("sf", "terra")
missing <- spatial_pkgs[!vapply(spatial_pkgs, requireNamespace, logical(1), quietly = TRUE)]
if (length(missing)) {
  if (dir.exists("/content")) {
    system("apt-get update -qq", ignore.stdout=TRUE, ignore.stderr=TRUE)
    system("apt-get install -y -qq libgdal-dev libgeos-dev libproj-dev libudunits2-dev", ignore.stdout=TRUE, ignore.stderr=TRUE)
  }
  install.packages(missing, dependencies=c("Depends", "Imports", "LinkingTo"), Ncpus=2)
}
library(sf); library(terra)
unidades <- st_read(course_file("data/spatial/unidades_geologicas.geojson"), quiet=TRUE)
dem <- rast(course_file("data/spatial/dem_sintetico.tif"))
cat("✓ sf funciona | geometrías:", nrow(unidades), "\n")
cat("✓ terra funciona | raster:", ncol(dem), "x", nrow(dem), "\n")
```
