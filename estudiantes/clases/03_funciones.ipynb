## De una operación a una función

Una **función** reúne instrucciones bajo un nombre para reutilizarlas con distintas entradas. Ya hemos llamado funciones como `mean()`, `length()` y `paste()`. Ahora escribiremos las nuestras.

Comenzaremos con una tarea: convertir una profundidad de metros a centímetros.

**Entrada → instrucciones → salida:** un número en metros → multiplicar por 100 → un número en centímetros.


````r
profundidad_m <- 2.5
profundidad_m * 100
````

## Estructura básica: un argumento y un resultado

```r
nombre_funcion <- function(argumento) {
  resultado <- instrucciones_que_usan_el_argumento
  return(resultado)
}
```

| Parte | Qué hace |
|---|---|
| `nombre_funcion <-` | Guarda la función en un objeto con ese nombre. |
| `function(argumento)` | Declara la entrada que la función recibe. |
| `{ ... }` | Delimita el cuerpo: las instrucciones que ejecutará. |
| `return(resultado)` | Entrega el resultado y termina esa llamada. |

**Definir** la función guarda las instrucciones; **llamarla** las ejecuta. El argumento es el nombre usado dentro de la definición; al llamar la función proporcionamos un valor para él.


````r
metros_a_cm <- function(metros) {
  centimetros <- metros * 100
  return(centimetros)
}
````

````r
metros_a_cm(2.5)
````

````r
metros_a_cm(0.8)
````

### Guardar y reutilizar el resultado

La salida es un objeto de R: puede guardarse, consultarse y participar en otras operaciones. `return()` entrega un valor; `print()` o `cat()` sirven para mostrarlo. Una función no necesita imprimirlo para que podamos usarlo.


````r
profundidad_cm <- metros_a_cm(1.2)
profundidad_cm
````

````r
profundidad_cm + 5
````

### Variables dentro y fuera de la función

`metros` y `centimetros` son nombres locales a esta llamada. Una asignación con `<-` dentro de la función no reemplaza un objeto del mismo nombre fuera de ella. Conviene pasar las entradas como argumentos para poder entender la función leyendo su definición.

## Varios argumentos: calcular porosidad

La porosidad porcentual se calcula como `(volumen de vacíos / volumen total) * 100`. Ambos volúmenes deben estar en la misma unidad.

```r
nombre_funcion <- function(entrada_1, entrada_2) {
  resultado <- operacion_con_ambas_entradas
  return(resultado)
}
```

En esta primera versión suponemos volúmenes válidos: total positivo y vacíos entre cero y el total. Más adelante comprobaremos esas condiciones.


````r
calcular_porosidad <- function(volumen_vacios, volumen_total) {
  porosidad_pct <- (volumen_vacios / volumen_total) * 100
  return(porosidad_pct)
}
````

````r
calcular_porosidad(18, 120)
````

Sin nombres, los valores se asocian según su posición. Con nombres podemos hacer explícito qué representa cada uno, incluso al cambiar el orden. Para funciones científicas, nombrar argumentos ayuda a evitar confundir medidas.


````r
calcular_porosidad(volumen_total = 120, volumen_vacios = 18)
````

## Decisiones dentro de una función: `if`, `else if` y `else`

La función siguiente recibe **un único porcentaje** y devuelve una etiqueta. Retomamos los límites didácticos del módulo 02: baja si es menor de 10 %, media entre 10 y 20 % incluidos, alta si supera 20 %.

Primero revisamos el ausente: `is.na()` devuelve un lógico, mientras que una comparación con `NA` no permite tomar la decisión. `if` necesita una condición de longitud uno.


````r
clasificar_porosidad <- function(porosidad_pct) {
  if (is.na(porosidad_pct)) {
    categoria <- "sin dato"
  } else if (porosidad_pct < 10) {
    categoria <- "baja"
  } else if (porosidad_pct <= 20) {
    categoria <- "media"
  } else {
    categoria <- "alta"
  }
  return(categoria)
}
````

````r
clasificar_porosidad(8)
````

````r
clasificar_porosidad(10)
````

````r
clasificar_porosidad(20)
````

````r
clasificar_porosidad(25)
````

````r
clasificar_porosidad(NA)
````

## Comprobar entradas y terminar antes de tiempo

La versión anterior presupone entradas válidas. Ahora una función de porosidad comprobará tipo, longitud y dominio.

- `stop("mensaje")` detiene una llamada con un error explicativo cuando la entrada es inválida.
- `return(NA_real_)` devuelve un ausente numérico cuando falta una medida.
- Un `return()` dentro de una rama termina toda la llamada, no solo esa rama.
- Las comprobaciones se hacen en orden: antes de comparar números, revisamos que las entradas sean numéricas y tengan longitud uno.

No se convierte un ausente en cero: cero vacíos y volumen desconocido representan situaciones diferentes.


````r
porosidad_segura <- function(volumen_vacios, volumen_total) {
  if (!is.numeric(volumen_vacios) || !is.numeric(volumen_total)) {
    stop("Los volúmenes deben ser numéricos.")
  }
  if (length(volumen_vacios) != 1 || length(volumen_total) != 1) {
    stop("Proporciona un solo valor para cada volumen.")
  }
  if (is.na(volumen_vacios) || is.na(volumen_total)) {
    return(NA_real_)
  }
  if (!is.finite(volumen_vacios) || !is.finite(volumen_total)) {
    stop("Los volúmenes deben ser finitos.")
  }
  if (volumen_total <= 0 || volumen_vacios < 0 || volumen_vacios > volumen_total) {
    stop("Se requiere total > 0 y vacíos entre 0 y el total.")
  }
  return((volumen_vacios / volumen_total) * 100)
}
````

````r
porosidad_segura(18, 120)
````

````r
porosidad_segura(NA_real_, 120)
````

## Una función puede recibir un vector y devolver un número

Recibir un vector no obliga a devolver un vector. Contaremos cuántos registros no ausentes indican presencia, recorriendo los conteos con `for`, como en el módulo 02.

`next` omite el registro ausente. Una abundancia de cero es un registro conocido de ausencia. Se suponen conteos no negativos.


````r
contar_presencias <- function(conteos) {
  presencias <- 0
  for (conteo in conteos) {
    if (is.na(conteo)) {
      next
    }
    if (conteo > 0) {
      presencias <- presencias + 1
    }
  }
  return(presencias)
}
````

````r
conteos_parcelas <- c(12, 0, NA, 8, 0, 5)
contar_presencias(conteos_parcelas)
````

## Reutilizar una función dentro de un ciclo

Para clasificar varias porosidades, llamaremos `clasificar_porosidad()` una vez por observación. `character(length(...))` prepara el vector de salida. `seq_along()` construye las posiciones que recorrerá `for`, incluso si el vector está vacío.

La función externa recibe un vector y devuelve otro de la misma longitud. La función interna sigue recibiendo un único porcentaje.


````r
clasificar_muestras <- function(porosidades) {
  categorias <- character(length(porosidades))
  for (i in seq_along(porosidades)) {
    categorias[i] <- clasificar_porosidad(porosidades[i])
  }
  return(categorias)
}
````

````r
clasificar_muestras(c(8, 10, 20, 25, NA))
````

## Incorporar `while`: contar pasos de una simulación

En una simulación didáctica avanzamos desde 0 m hasta alcanzar o superar una profundidad objetivo. El objetivo debe ser no negativo y el paso debe ser positivo; ambos son números individuales y finitos.

El argumento `paso_m` tiene valor por defecto. El ciclo actualiza la profundidad y cuenta pasos. Si el paso no divide exactamente el objetivo, la profundidad final puede superarlo. Con objetivo cero, el ciclo no se ejecuta.


````r
contar_pasos <- function(objetivo_m, paso_m = 0.5) {
  if (!is.numeric(objetivo_m) || !is.numeric(paso_m)) {
    stop("Objetivo y paso deben ser numéricos.")
  }
  if (length(objetivo_m) != 1 || length(paso_m) != 1) {
    stop("Objetivo y paso deben tener longitud uno.")
  }
  if (!is.finite(objetivo_m) || !is.finite(paso_m)) {
    stop("Objetivo y paso deben ser finitos y no ausentes.")
  }
  if (objetivo_m < 0 || paso_m <= 0) {
    stop("El objetivo debe ser no negativo y el paso positivo.")
  }
  profundidad_m <- 0
  pasos <- 0
  while (profundidad_m < objetivo_m) {
    profundidad_m <- profundidad_m + paso_m
    pasos <- pasos + 1
  }
  return(pasos)
}
````

````r
contar_pasos(3)
````

````r
contar_pasos(3, paso_m = 0.8)
````

````r
contar_pasos(0)
````

## Una matriz como entrada: ciclos anidados

Cada fila representa una parcela y cada columna una especie. Usaremos los dos ciclos del módulo 02 para calcular un total de individuos. En este ejemplo todos los conteos están presentes.

**Predice:** ¿cuántas sumas se ejecutan? ¿Qué cambiarías para devolver un total por parcela?


````r
sumar_abundancias <- function(conteos) {
  total <- 0
  for (i in seq_len(nrow(conteos))) {
    for (j in seq_len(ncol(conteos))) {
      total <- total + conteos[i, j]
    }
  }
  return(total)
}
````

````r
abundancias <- matrix(c(4, 2, 0, 3, 5, 1), nrow = 2, byrow = TRUE)
sumar_abundancias(abundancias)
````

## Una tabla como entrada y una lista como salida

Una función devuelve **un objeto**. Si necesitamos varias piezas de información, ese objeto puede ser una lista con nombres. Aquí la entrada es una tabla con columnas `codigo` y `porosidad_pct`.

Usaremos `$` para acceder a una columna, contaremos los datos conocidos y calcularemos su media. Si todos están ausentes, devolveremos `NA_real_` en lugar de calcular una media sin observaciones.


````r
resumir_muestras <- function(tabla) {
  porosidades <- tabla$porosidad_pct
  n_validos <- sum(!is.na(porosidades))
  if (n_validos == 0) {
    media_pct <- NA_real_
  } else {
    media_pct <- mean(porosidades, na.rm = TRUE)
  }
  resultado <- list(
    n_muestras = nrow(tabla),
    n_validos = n_validos,
    porosidad_media_pct = media_pct,
    categorias = clasificar_muestras(porosidades)
  )
  return(resultado)
}
````

````r
muestras <- data.frame(
  codigo = c("G1", "G2", "G3", "G4"),
  porosidad_pct = c(8, 15, NA, 25)
)
muestras
````

````r
resumen_muestras <- resumir_muestras(muestras)
resumen_muestras
````

````r
resumen_muestras$porosidad_media_pct
````
