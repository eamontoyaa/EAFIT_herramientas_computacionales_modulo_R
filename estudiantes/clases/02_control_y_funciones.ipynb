## `if` y `else`

- `if` necesita una condición de longitud uno cuyo valor sea `TRUE` o `FALSE`.
- La rama de `if` se ejecuta cuando la condición es verdadera.
- `else` indica qué hacer cuando es falsa; esta rama es opcional.
- La asignación usa `<-`; la comparación de igualdad usa `==`.
- Escribe `else` junto a la llave que cierra la rama anterior.

**Estructura básica:**

```r
if (condicion) {
  # Instrucciones si la condición es TRUE
} else {
  # Instrucciones si la condición es FALSE
}
```


````r
if (abundancia > umbral) {
  categoria <- "supera el umbral"
} else {
  categoria <- "no supera el umbral"
}
categoria
````

## Varias alternativas y valores ausentes

- `else if` permite evaluar otra condición cuando las anteriores son falsas.
- Se ejecuta únicamente la primera rama cuya condición sea verdadera.
- `else` recoge los casos que no cumplen ninguna condición anterior.
- Revisa primero si falta el dato: `if (NA)` no permite decidir una rama.

**Estructura básica:**

```r
if (condicion_1) {
  # Primera alternativa
} else if (condicion_2) {
  # Segunda alternativa
} else {
  # Casos restantes
}
```

En esta clasificación didáctica de porosidad:

- **Baja:** menos de 10 %.
- **Media:** de 10 a 20 %, incluidos ambos límites.
- **Alta:** más de 20 %.

**Predice:** ¿qué pasa con 10, 20 y `NA`?


````r
porosidad_pct <- 20
if (is.na(porosidad_pct)) {
  clase <- "sin dato"
} else if (porosidad_pct < 10) {
  clase <- "baja"
} else if (porosidad_pct <= 20) {
  clase <- "media"
} else {
  clase <- "alta"
}
clase
````

## Condiciones escalares y vectoriales

- `&` y `|` combinan condiciones elemento a elemento.
- `&&` y `||` operan sobre condiciones escalares y pueden evitar evaluar la segunda condición.
- `any()` comprueba si al menos un elemento cumple la condición.
- `all()` comprueba si todos los elementos cumplen la condición.
- `ifelse()` produce una elección por elemento.


````r
riqueza <- c(12, 25, 18, 30)
riqueza >= 15 & riqueza <= 25
````

````r
any(riqueza > 25)
````

````r
any(riqueza > 35)
````

````r
all(riqueza >= 0)
````

````r
all(riqueza >= 15)
````

````r
n_parcelas <- length(riqueza)
if (n_parcelas > 0 && mean(riqueza) > 20) {
  mensaje <- "Promedio mayor que 20 especies"
} else {
  mensaje <- "No se cumple la condición"
}
mensaje
````

- `ifelse()` evalúa una condición lógica sobre un vector elemento por elemento.
- **Uso principal:** crear nuevas variables o transformar columnas en un `data.frame` basándose en condiciones.
- **Manejo de `NA`:** si un elemento de la condición es `NA`, el resultado correspondiente será `NA`.

**Sintaxis básica:**

```r
ifelse(condicion, valor_si_true, valor_si_false)
```


````r
riqueza
````

````r
ifelse(riqueza >= 20, "alta", "baja")
````

## `for`: recorrer valores

- `for` repite su cuerpo para cada elemento de un vector.
- La variable del ciclo toma el valor del elemento actual en cada iteración.
- Inicializa el acumulador antes de comenzar.

**Estructura básica:**

```r
for (elemento in valores) {
  # Instrucciones que usan el elemento actual
}
```


````r
nivel_freatico <- c(0.5, 2.3, 0.8, 4.1)
i <- 1
for (nf in nivel_freatico) {
    cat("Nivel freático en interacion", i, ":", nf, "m\n")
    i <- i + 1
}
````

Traza en una tabla el espesor actual y el acumulado después de cada iteración.

````r
espesores_m <- c(1.2, 0.8, 2.1, 1.5)
total_m <- 0
for (espesor in espesores_m) {
  total_m <- total_m + espesor
}
total_m
````

## Recorrer posiciones y reservar espacio

- Usa `seq_along()` cuando necesites la posición de cada elemento.
- Reserva el vector de salida antes del ciclo para evitar ampliarlo en cada vuelta.
- Guarda cada resultado en su posición para relacionarlo con la entrada correspondiente.

**Estructura básica:**

```r
salida <- numeric(length(valores))
for (i in seq_along(valores)) {
  salida[i] <- valores[i] * 100
}
```

Este esquema reserva una salida numérica y guarda cada valor multiplicado por 100.


````r
espesores_cm <- numeric(length(espesores_m))
for (i in seq_along(espesores_m)) {
  espesores_cm[i] <- espesores_m[i] * 100
  cat("Espesor en iteración", i, ":", espesores_m[i], "m ->", espesores_cm[i], "cm\n")
}
espesores_cm
````

````r
espesores_m * 100
````

## Dos índices: ciclos anidados

- Las filas son parcelas y las columnas son especies.
- El ciclo externo recorre cada parcela mediante `i`.
- El ciclo interno recorre sus especies mediante `j`.
- Por cada iteración del ciclo externo, se ejecuta todo el ciclo interno.

**Estructura básica:**

```r
for (i in seq_len(nrow(matriz))) {
  for (j in seq_len(ncol(matriz))) {
    # Instrucciones que usan matriz[i, j]
  }
}
```

Explica qué significan `i` y `j` y cuántas veces se ejecuta la suma del ejemplo.


````r
seq_len(5)
````

````r
conteos <- matrix(c(4, 2, 0, 3, 5, 1), nrow = 2, byrow = TRUE)
total <- 0
for (i in seq_len(nrow(conteos))) {
  for (j in seq_len(ncol(conteos))) {
    total <- total + conteos[i, j]
  }
}
total
````

````r
sum(conteos)
````

## `while`: repetir mientras se cumpla una condición

- `while` revisa la condición antes de cada iteración.
- El cuerpo puede no ejecutarse si la condición es falsa desde el inicio.
- Inicializa la variable de control antes del ciclo.
- Actualiza la variable de control para que el ciclo termine.

**Estructura básica:**

```r
# Inicializar la variable de control
while (condicion) {
  # Instrucciones que se repiten
  # Actualizar la variable de control
}
```

En una simulación didáctica se avanza 0.5 m por paso hasta 3 m.


````r
profundidad_m <- 0
paso_m <- 0.5
pasos <- 0
while (profundidad_m < 3) {
  profundidad_m <- profundidad_m + paso_m
  pasos <- pasos + 1
}
cat("Se alcanzó una profundidad de", profundidad_m, "m en", pasos, "pasos de", paso_m, "m cada uno.\n")
````

## `repeat` y `break`

- `repeat` comienza sin comprobar una condición.
- `break` termina el ciclo más cercano que lo contiene.
- Incluye una condición de salida para evitar que el ciclo se repita indefinidamente.
- Un máximo de iteraciones permite terminar incluso si no se logra la tolerancia.

**Estructura básica:**

```r
repeat {
  # Instrucciones y actualización de valores
  if (condicion_de_salida) {
    break
  }
}
```

Esta aproximación numérica se detiene al acercarse al objetivo o al alcanzar un máximo de iteraciones.


````r
estimacion <- 0
objetivo <- 10
tolerancia <- 0.01
iteracion <- 0
max_iteraciones <- 20
repeat {
  estimacion <- (estimacion + objetivo) / 2
  iteracion <- iteracion + 1
  if (abs(estimacion - objetivo) < tolerancia || iteracion >= max_iteraciones) {
    break
  }
}
print(c(estimacion = estimacion, iteraciones = iteracion))
````

## `next`: omitir una iteración

- `next` omite las instrucciones restantes de la iteración actual y pasa a la siguiente.
- Omitir un ausente no significa convertirlo en cero.
- Cuenta también cuántas observaciones aportan al total.

**Estructura básica:**

```r
for (elemento in valores) {
  if (condicion_para_omitir) {
    next
  }
  # Procesar el elemento si no se omitió
}
```

El registro de abundancia del ejemplo contiene un ausente.


````r
abundancias <- c(12, NA, 0, 8, 5)
total_observado <- 0
n_observados <- 0
for (abundancia in abundancias) {
  if (is.na(abundancia)) {
    next
  }
  total_observado <- total_observado + abundancia
  n_observados <- n_observados + 1
}
print(c(total = total_observado, observaciones = n_observados))
````
