## Tipos y operaciones

El tipo de un dato determina qué operaciones tienen sentido. En R, los booleanos son `logical` (`TRUE` y `FALSE`); los enteros pueden escribirse con `L` y los reales con decimales. Un carácter y una cadena de varios caracteres son ambos de tipo `character`.

### Operaciones numéricas

| Operador en R | Operación | Ejemplo |
|---|---|---|
| `+`, `-` | Suma y resta | `18 + 7`, `18 - 7` |
| `*`, `/` | Producto y división | `18 * 7`, `18 / 7` |
| `%/%` | Cociente entero | `18 %/% 7` |
| `%%` | Residuo | `18 %% 7` |
| `^` | Potencia | `3 ^ 4` |

Si se reparten 18 muestras en grupos de 7, el cociente cuenta grupos completos y el residuo cuenta las muestras restantes.


```r
18 + 7
```

```r
18 - 7
```

```r
18 * 7
```

```r
18 / 7
```

```r
18 %/% 7
```

```r
18 %% 7
```

```r
3 ^ 4
```

### Operaciones con texto

- En R se concatena con `paste()` o `paste0()`; `+` se reserva para operaciones numéricas. 
- `substr()` extrae caracteres entre dos posiciones, contando desde 1. `substring()` también extrae texto; 
- `regexpr()` localiza la primera coincidencia y devuelve `-1` si no la encuentra. `gregexpr()` encuentra todas las coincidencias.
- `cat()` escribe un mensaje directamente en la salida.


```r
codigo_muestra <- paste0("GEO", "-", "014")  # No añade espacios entre los elementos
codigo_muestra
```

```r
nchar(codigo_muestra)
```

```r
paste("Muestra", codigo_muestra)  # Añade un espacio entre los elementos
```

```r
substr(codigo_muestra, start = 1, stop = 3)
```

```r
regexpr("-", codigo_muestra, fixed = TRUE)
```

```r
substr(
    codigo_muestra,
    start = regexpr("-", codigo_muestra, fixed = TRUE) + 1,
    stop = nchar(codigo_muestra)
)
```

```r
regexpr("X", codigo_muestra, fixed = TRUE)
```

```r
substring(codigo_muestra, first = 5)
```

```r
cat("Identificador:", codigo_muestra, "\n")
```

```r
as.integer(gregexpr("A", "A1-A2-A3", fixed = TRUE)[[1]])
```

### Comparaciones

Los operadores `>`, `>=`, `<`, `<=`, `==` y `!=` producen valores lógicos. Distingue igualdad (`==`) de asignación (`<-`). Comprueba los límites: una muestra exactamente en el umbral puede cambiar de grupo al pasar de `>` a `>=`.


```r
densidad <- 2.7
densidad > 2.7
```

```r
densidad >= 2.7
```

```r
densidad < 2.7
```

```r
densidad <= 2.7
```

```r
densidad == 2.7
```

```r
densidad != 2.7
```

### Conectivos lógicos

La conjunción exige que ambas condiciones se cumplan; la disyunción exige al menos una; `!` niega una condición. Para decisiones sobre un solo valor se usan `&&` y `||`. En vectores, `&` y `|` operan elemento a elemento, como veremos al seleccionar observaciones.


```r
densidad >= 2.5 && densidad <= 2.8
```

```r
densidad < 2.5 || densidad > 2.8
```

```r
!(densidad >= 2.5 && densidad <= 2.8)
```

## Precedencia de operadores en R

Los paréntesis explicitan qué parte se calcula primero. Para los operadores usados en esta sesión, el orden de mayor a menor prioridad es:

| Prioridad | Operadores |
|---|---|
| 1 | `^` |
| 2 | `-` y `+` unarios (signo) |
| 3 | `:` |
| 4 | `%/%`, `%%`, `%in%`, `%*%`, `%o%` |
| 5 | `*`, `/` |
| 6 | `+`, `-` binarios (suma y resta) |
| 7 | `<`, `>`, `<=`, `>=`, `==`, `!=` |
| 8 | `!` |
| 9 | `&`, `&&` |
| 10 | `|`, `||` |
| 11 | `<-` |

Esta tabla resume las reglas de R. Con igual prioridad se evalúa normalmente de izquierda a derecha, pero las potencias se agrupan de derecha a izquierda. La negación se aplica antes que la conjunción y la disyunción.

Observa cómo los paréntesis cambian una expresión aritmética.


```r
4 + 3 * 2
```

```r
(4 + 3) * 2
```

### Signo, potencia y asociatividad

Un signo negativo no forma parte de la base de una potencia salvo que lo agrupes. En potencias encadenadas, haz explícita la agrupación para evitar confusiones.


```r
-4 ^ 2
```

```r
(-4) ^ 2
```

```r
2 ^ 3 ^ 2
```

```r
(2 ^ 3) ^ 2
```

```r
2 ^ (3 ^ 2)
```

### Cocientes, secuencias y lógica

En R, los operadores especiales como `%%` se evalúan antes que `*`. `:` se evalúa antes que la suma. Una conjunción tiene prioridad sobre una disyunción. Los siguientes pares permiten observar estas diferencias.


```r
4 * 7 %% 3
```

```r
(4 * 7) %% 3
```

```r
1:4 + 1
```

```r
1:(4 + 1)
```

```r
!FALSE && FALSE
```

```r
!(FALSE && FALSE)
```

```r
TRUE || FALSE && FALSE
```

```r
(TRUE || FALSE) && FALSE
```

### Una fórmula con unidades

Una muestra sintética tiene 18 cm³ de vacíos y 120 cm³ de volumen total. Al escribir la porosidad porcentual, los paréntesis ayudan a comunicar el cociente que se transforma a porcentaje.


```r
volumen_vacios_cm3 <- 18
volumen_total_cm3 <- 120
porosidad_calculada_pct <- (volumen_vacios_cm3 / volumen_total_cm3) * 100
porosidad_calculada_pct
```

`NA` representa una observación desconocida, `NaN` un resultado numérico indefinido e `Inf` un infinito. Usa `is.na()` para detectar ausentes, no `x == NA`. Omitir ausentes en un resumen debe ser una decisión explícita.

## Matrices y atributos

Una matriz tiene filas y columnas, con un mismo tipo de dato. Por defecto se llena por columnas; `byrow = TRUE` permite llenarla por filas. `dim()` da sus dimensiones y `attributes()` muestra atributos como dimensiones y nombres.

Los datos siguientes son abundancias de tres especies en dos parcelas.


```r
abundancias <- matrix(c(12, 4, 0, 8, 6, 3), nrow = 2, byrow = TRUE)
dimnames(abundancias) <- list(c("P1", "P2"), c("especie_a", "especie_b", "especie_c"))
abundancias
```

```r
dim(abundancias)
```

```r
attributes(abundancias)
```

```r
cbind(abundancias, especie_d = c(2, 1))
```

```r
rbind(abundancias, P3 = c(5, 7, 2))
```

## Operaciones con vectores y matrices

`+`, `-`, `*` y `/` actúan elemento a elemento. Compara los mismos valores como vectores y como matrices.


```r
vector_a <- c(2, 4, 6)
vector_b <- c(1, 2, 3)
vector_a + vector_b
```

```r
vector_a - vector_b
```

```r
vector_a * vector_b
```

```r
vector_a / vector_b
```

```r
matriz_a <- matrix(c(2, 4, 6, 8), nrow = 2, byrow = TRUE)
matriz_b <- matrix(c(1, 2, 3, 4), nrow = 2, byrow = TRUE)
matriz_a + matriz_b
```

```r
matriz_a - matriz_b
```

```r
matriz_a * matriz_b
```

```r
matriz_a / matriz_b
```

### Productos entre objetos

`%*%` calcula el producto matricial: las columnas del primer objeto deben coincidir con las filas del segundo. Con dos vectores de igual longitud calcula su producto punto. `%o%` combina cada elemento del primer vector con todos los del segundo.


```r
vector_a %*% vector_b
```

```r
matriz_a %*% matriz_b
```

```r
vector_a %o% vector_b
```

### Aplicación: biomasa por parcela

Cada fila contiene abundancias de tres especies; cada especie tiene una masa media sintética en gramos. El producto matricial estima la biomasa total de cada parcela. El producto elemento a elemento tendría otro significado.


```r
conteos_parcelas <- matrix(c(4, 2, 1, 3, 0, 5), nrow = 2, byrow = TRUE)
masa_media_g <- c(2, 5, 3)
conteos_parcelas %*% masa_media_g
```

## Factores para categorías

Un factor representa categorías mediante niveles. Su orden se declara cuando tiene significado. Los códigos internos de un factor no son mediciones: `as.numeric(factor)` devuelve esos códigos.


```r
alteracion <- factor(c("alta", "baja", "media", "baja"),
                     levels = c("baja", "media", "alta"), ordered = TRUE)
alteracion
```

```r
levels(alteracion)
```

```r
class(alteracion)
```

```r
as.numeric(alteracion)
```

### Etiquetas de un factor

```r
as.character(alteracion)
```

## Selección con índices y condiciones

R comienza a contar posiciones en 1. Un índice negativo excluye posiciones. Un vector lógico selecciona donde hay `TRUE`; combina condiciones con `&` (y), `|` (o) y `!` (negación).

**Predice:** ¿qué estratos tienen un espesor entre 1 y 2 m?


```r
espesor_m[1]
```

```r
espesor_m[c(1, 3)]
```

```r
espesor_m[-2]
```

```r
espesor_m["E3"]
```

```r
espesor_m[espesor_m >= 1 & espesor_m <= 2]
```

```r
porosidad_pct[!is.na(porosidad_pct) & porosidad_pct > 16]
```

## Selección en matrices, listas y tablas

Usa `[fila, columna]` en matrices y tablas. En una lista, `[` conserva una lista; `[[` extrae un elemento y `$` lo busca por nombre. `drop = FALSE` conserva las dimensiones al seleccionar una sola fila o columna.


```r
abundancias[1, ]
```

```r
abundancias[, 2, drop = FALSE]
```

```r
sitio["abundancias"]
```

```r
sitio[["abundancias"]]
```

```r
sitio$ambiente$elevacion_m
```

```r
geo[geo$fracturas > 5, c("id", "fracturas")]
```

```r
geo[["densidad_g_cm3"]]
```

## Localizar y ordenar

`which()` devuelve posiciones que cumplen una condición. `order()` devuelve posiciones de ordenamiento: puedes usarlas para reorganizar una tabla completa. `rev()` invierte el orden actual.


```r
which(geo$fracturas > 5)
```

```r
order(geo$fracturas, decreasing = TRUE)
```

```r
geo[order(geo$fracturas, decreasing = TRUE), ]
```

```r
rev(espesor_m)
```

## Pertenencia y conjuntos

Dos parcelas registran las especies que aparecen abajo. `%in%` evalúa pertenencia; `union()`, `intersect()` y `setdiff()` comparan conjuntos sin conservar repeticiones.

**Interpretación:** ¿qué información de abundancia se pierde al convertir registros en un conjunto?


```r
parcela_a <- c("roble", "palma", "helecho", "roble")
parcela_b <- c("palma", "musgo", "helecho")
parcela_a %in% parcela_b
```

```r
union(parcela_a, parcela_b)
```

```r
intersect(parcela_a, parcela_b)
```

```r
setdiff(parcela_a, parcela_b)
```

```r
setdiff(parcela_b, parcela_a)
```
