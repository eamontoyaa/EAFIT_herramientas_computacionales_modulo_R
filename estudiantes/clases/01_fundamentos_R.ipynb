## Tipos y clases

`class()`, `typeof()` y `mode()` describen aspectos distintos de un objeto; no son tres nombres para la misma consulta.

| Función | Qué describe | Para qué sirve |
|---|---|---|
| `class(x)` | La clase del objeto, como `numeric`, `integer` o `factor`. | Indica cómo se interpreta el objeto y ayuda a R a elegir cómo tratarlo en funciones como `print()` o `summary()`. |
| `typeof(x)` | El tipo interno con el que R almacena el objeto, como `double`, `integer` o `character`. | Distingue la representación interna de los datos. |
| `mode(x)` | Una clasificación más general del tipo interno. Agrupa `double` e `integer` bajo `numeric`. | Permite consultar esa categoría amplia sin distinguir enteros de números de doble precisión. |

Por ejemplo, `12.5` tiene clase `numeric`, tipo `double` y modo `numeric`. En cambio, `12L` tiene clase y tipo `integer`, pero su modo también es `numeric`. La letra `L` solicita almacenamiento entero.


```r
class(profundidad_m)
```

```r
typeof(profundidad_m)
```

```r
mode(profundidad_m)
```

```r
class(12L)
```

```r
typeof(12L)
```

```r
mode(12L)
```

### Reconocer tipos y clases con `is.*()`

Estas funciones responden `TRUE` o `FALSE`: comprueban una propiedad, sin convertir el objeto.

| Función | Qué comprueba |
|---|---|
| `is.integer(x)` | Si el almacenamiento es entero. |
| `is.double(x)` | Si el almacenamiento es de doble precisión. |
| `is.numeric(x)` | Si es un objeto numérico de tipo entero o doble, como los vectores de números de estos ejemplos; no considera numéricos a los factores. |
| `is.character(x)` | Si es texto. |
| `is.logical(x)` | Si es de tipo lógico. |
| `is.complex(x)` | Si es de tipo complejo. |
| `is.factor(x)` | Si es un factor. |

**Un valor sin decimales no necesariamente tiene almacenamiento entero:** `12` se almacena como `double`, mientras que `12L` se almacena como `integer`. `is.integer()` no pregunta si el valor carece de parte fraccionaria.


```r
is.integer(12)
```

```r
is.integer(12L)
```

```r
is.double(12)
```

```r
is.double(12L)
```

```r
is.numeric(profundidad_m)
```

```r
is.numeric(12L)
```

Las comillas hacen que `"12"` sea texto, aunque sus caracteres sean dígitos. Un valor lógico tampoco es numérico para `is.numeric()`.

## Tipos y operaciones

El tipo de un dato determina qué operaciones tienen sentido. En R, los booleanos son `logical` (`TRUE` y `FALSE`); los enteros pueden escribirse con `L` y los reales con decimales. Un carácter y una cadena de varios caracteres son ambos de tipo `character`.

### Operaciones numéricas

| Operador en R | Operación | Ejemplo |
|---|---|---|
| `+`, `-` | Suma y resta | `18 + 7`, `18 - 7` |
| `*`, `/` | Producto y división | `18 * 7`, `18 / 7` |
| `%/%` | Cociente entero | `18 %/% 7` |
| `%%` | Residuo | `18 %% 7` |
| `^` | Potencia | `3 ^ 4` |

Si se reparten 18 muestras en grupos de 7, el cociente cuenta grupos completos y el residuo cuenta las muestras restantes.


```r
18 + 7
```

```r
18 - 7
```

```r
18 * 7
```

```r
18 / 7
```

```r
18 %/% 7
```

```r
18 %% 7
```

```r
3 ^ 4
```

### Operaciones con texto

- En R se concatena con `paste()` o `paste0()`; `+` se reserva para operaciones numéricas. 
- `substr()` extrae caracteres entre dos posiciones, contando desde 1. `substring()` también extrae texto; 
- `regexpr()` localiza la primera coincidencia y devuelve `-1` si no la encuentra. `gregexpr()` encuentra todas las coincidencias.
- `cat()` escribe un mensaje directamente en la salida.

### Comparaciones

- Los operadores `>`, `>=`, `<`, `<=`, `==` y `!=` producen valores lógicos.
- Distingue igualdad (`==`) de asignación (`<-`).
- Comprueba los límites: una muestra exactamente en el umbral puede cambiar de grupo al pasar de `>` a `>=`.


```r
densidad <- 2.7
densidad > 2.7
```

```r
densidad >= 2.7
```

```r
densidad < 2.7
```

```r
densidad <= 2.7
```

```r
densidad == 2.7
```

```r
densidad != 2.7
```

### Conectivos lógicos

- La conjunción exige que ambas condiciones se cumplan.
- La disyunción exige que se cumpla al menos una condición.
- `!` niega una condición.
- Para decisiones sobre un solo valor se usan `&&` y `||`.
- En vectores, `&` y `|` operan elemento a elemento, como veremos al seleccionar observaciones.


```r
densidad >= 2.5 && densidad <= 2.8
```

```r
densidad < 2.5 || densidad > 2.8
```

```r
!(densidad >= 2.5 && densidad <= 2.8)
```

## Precedencia de operadores en R

Los paréntesis explicitan qué parte se calcula primero. Para los operadores usados en esta sesión, el orden de mayor a menor prioridad es:

| Prioridad | Operadores |
|---|---|
| 1 | `^` |
| 2 | `-` y `+` unarios (signo) |
| 3 | `:` |
| 4 | `%/%`, `%%`, `%in%`, `%*%`, `%o%` |
| 5 | `*`, `/` |
| 6 | `+`, `-` binarios (suma y resta) |
| 7 | `<`, `>`, `<=`, `>=`, `==`, `!=` |
| 8 | `!` |
| 9 | `&`, `&&` |
| 10 | `\|`, `\|\|` |
| 11 | `<-` |

Esta tabla resume las reglas de R. Con igual prioridad se evalúa normalmente de izquierda a derecha, pero las potencias se agrupan de derecha a izquierda. La negación se aplica antes que la conjunción y la disyunción.

Observa cómo los paréntesis cambian una expresión aritmética.


```r
4 + 3 * 2
```

```r
(4 + 3) * 2
```

### Signo, potencia y asociatividad

- Un signo negativo no forma parte de la base de una potencia salvo que lo agrupes.
- En potencias encadenadas, haz explícita la agrupación para evitar confusiones.


```r
-4 ^ 2
```

```r
(-4) ^ 2
```

```r
2 ^ 3 ^ 2
```

```r
(2 ^ 3) ^ 2
```

```r
2 ^ (3 ^ 2)
```

### Cocientes, secuencias y lógica

- En R, los operadores especiales como `%%` se evalúan antes que `*`.
- `:` se evalúa antes que la suma.
- Una conjunción tiene prioridad sobre una disyunción.

Los siguientes pares permiten observar estas diferencias.


```r
4 * 7 %% 3
```

```r
(4 * 7) %% 3
```

```r
1:4 + 1
```

```r
1:(4 + 1)
```

```r
!FALSE && FALSE
```

```r
!(FALSE && FALSE)
```

```r
TRUE || FALSE && FALSE
```

```r
(TRUE || FALSE) && FALSE
```

### Productos entre objetos

```r
vector_a %*% vector_b
```

Cada fila contiene abundancias de tres especies; cada especie tiene una masa media sintética en gramos. El producto matricial estima la biomasa total de cada parcela. El producto elemento a elemento tendría otro significado.

## Factores para categorías

Estructura para representar **datos categóricos**. Almacena internamente números enteros asociados a etiquetas llamadas **niveles** (`levels`).

- Un factor representa categorías mediante niveles.
- Su orden se declara cuando tiene significado.
- Los códigos internos de un factor no son mediciones: `as.numeric(factor)` devuelve esos códigos.


```r
# 1. Factor Nominal
tipo_suelo <- factor(c("Arcilla", "Arena", "Limo", "Arcilla"))
print(tipo_suelo)
```

```r
levels(tipo_suelo)
```

```r
class(tipo_suelo)
```

```r
as.numeric(tipo_suelo)
```

```r
# 2. Factor Ordinal (con parámetro ordered = TRUE)
amenaza <- factor(c("Baja", "Alta", "Media", "Alta"), 
                  levels = c("Baja", "Media", "Alta"), 
                  ordered = TRUE)
print(amenaza)
```

```r
class(amenaza)
```

```r
as.numeric(amenaza)
```

### Etiquetas de un factor

Para trabajar con el texto de sus categorías, convierte el factor con `as.character()`. Esto conserva las etiquetas, en lugar de devolver los códigos internos.


```r
as.character(amenaza)
```

```r
paste0("Amenaza: ", as.character(amenaza))
```

## Pertenencia y conjuntos

Dos parcelas registran las especies que aparecen abajo.

- `%in%` evalúa pertenencia.
- `union()`, `intersect()` y `setdiff()` comparan conjuntos sin conservar repeticiones.

```r
parcela_a <- c("roble", "palma", "helecho", "roble")
parcela_b <- c("palma", "musgo", "helecho")
parcela_a %in% parcela_b
```

```r
union(parcela_a, parcela_b)
```

```r
intersect(parcela_a, parcela_b)
```

```r
setdiff(parcela_a, parcela_b)
```

```r
setdiff(parcela_b, parcela_a)
```
